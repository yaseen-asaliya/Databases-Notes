## Concept 1 — WHERE and Comparison Operators

`WHERE` filters rows **before** SELECT runs. Only rows that satisfy the condition are returned.

```sql
SELECT title FROM film WHERE rental_rate = 4.99;
SELECT title FROM film WHERE length > 120;
SELECT title FROM film WHERE rating != 'G';
```

| Operator | Meaning |
|---|---|
| `=` | equal |
| `!=` or `<>` | not equal |
| `>` / `<` | greater / less than |
| `>=` / `<=` | greater or equal / less or equal |

## Concept 2 — AND, OR, NOT

Combine multiple conditions:

```sql
-- AND: both conditions must be true
SELECT title FROM film WHERE rating = 'PG' AND length > 100;

-- OR: at least one condition must be true
SELECT title FROM film WHERE rating = 'G' OR rating = 'PG';

-- NOT: inverts the condition
SELECT title FROM film WHERE NOT rating = 'R';
```

⚠️ Use parentheses when mixing AND and OR — AND has higher precedence than OR:
```sql
-- Without parens: AND binds first, giving wrong logic
WHERE rating = 'G' OR rating = 'PG' AND length > 100

-- Correct: use parens to control grouping
WHERE (rating = 'G' OR rating = 'PG') AND length > 100
```

## Concept 3 — IN and NOT IN

`IN` checks if a value matches any value in a list — cleaner than multiple OR conditions:

```sql
-- These two are equivalent:
WHERE rating = 'G' OR rating = 'PG' OR rating = 'PG-13'
WHERE rating IN ('G', 'PG', 'PG-13')
```

`NOT IN` excludes those values:
```sql
SELECT title FROM film WHERE rating NOT IN ('R', 'NC-17');
```

## Concept 4 — BETWEEN

`BETWEEN` filters for values within an inclusive range:

```sql
SELECT title FROM film WHERE length BETWEEN 90 AND 120;
-- equivalent to: WHERE length >= 90 AND length <= 120
```

Works on numbers, dates, and strings:
```sql
WHERE payment_date BETWEEN '2005-06-01' AND '2005-06-30'
```

## Concept 5 — LIKE (Pattern Matching)

`LIKE` matches string patterns using wildcards:

| Wildcard | Meaning |
|---|---|
| `%` | zero or more of any character |
| `_` | exactly one of any character |

```sql
WHERE title LIKE 'A%'        -- starts with A
WHERE title LIKE '%love%'    -- contains "love" anywhere
WHERE title LIKE '%ing'      -- ends with "ing"
WHERE title LIKE '_AT%'      -- second and third chars are A and T
```

`NOT LIKE` excludes matching rows:
```sql
WHERE title NOT LIKE 'A%'
```

## Concept 6 — IS NULL and IS NOT NULL

`NULL` means missing or unknown. You **cannot** compare NULL with `=` — it will never match:

```sql
-- WRONG — never returns rows:
WHERE original_language_id = NULL

-- CORRECT:
WHERE original_language_id IS NULL
WHERE original_language_id IS NOT NULL
```

NULL is not zero, not empty string — it means the value is absent.

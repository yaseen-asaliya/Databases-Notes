# Phase 15 — Views

A view is a saved SELECT query stored in the database with a name. You query it like a table.

```sql
CREATE VIEW view_name AS
SELECT ...;

SELECT * FROM view_name;

DROP VIEW view_name;
```

**Why use views:**
- Simplify complex joins — expose a clean interface to applications
- Security — grant access to a view without exposing underlying tables
- Reusability — write a complex query once, use it everywhere

**Key limitation:** Views don't store data — they re-run the query every time you SELECT from them. For performance, use materialized views (not available in MySQL — use a summary table instead).

**Updatable views:** A view is updatable (INSERT/UPDATE/DELETE) only if it maps 1:1 to a single table and has no GROUP BY, DISTINCT, aggregates, or JOINs.

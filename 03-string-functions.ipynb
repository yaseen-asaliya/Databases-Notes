# Phase 3 — String Functions

| Function | What it does |
|---|---|
| `CONCAT(a, b, ...)` | Join strings together |
| `CONCAT_WS(sep, a, b)` | Join with a separator |
| `UPPER(s)` / `LOWER(s)` | Change case |
| `LENGTH(s)` | Number of bytes |
| `CHAR_LENGTH(s)` | Number of characters (use for Unicode) |
| `TRIM(s)` | Remove leading and trailing spaces |
| `LTRIM(s)` / `RTRIM(s)` | Trim one side only |
| `SUBSTRING(s, pos, len)` | Extract part of a string (1-indexed) |
| `REPLACE(s, from, to)` | Replace all occurrences |
| `INSTR(s, substr)` | Position of first match (0 = not found) |
| `LEFT(s, n)` / `RIGHT(s, n)` | First / last N characters |
| `LPAD(s, len, pad)` / `RPAD(...)` | Pad string to a given length |

## CONCAT — build full names

```python
%%sql
-- Q1: Show each actor's full name as 'first_name last_name' from the actor table — label the column full_name

SELECT CONCAT(first_name, ' ', last_name) AS full_name FROM actor;
```

---
## CONCAT_WS — join with separator

`CONCAT_WS` skips NULL values automatically, unlike `CONCAT`.

```python
%%sql
-- Q2: Show each actor's name as 'LAST_NAME, first_name' using CONCAT_WS — from actor table — label the column full_name

SELECT CONCAT_WS(', ', first_name, last_name) AS full_name FROM actor;
```

---
## UPPER / LOWER / LENGTH

```python
%%sql
-- Q3: For each film, show the title in UPPERCASE, LOWERCASE, and its character length — from film table — label columns upper_title, lower_title, title_length

SELECT UPPER(title) AS upper_title, LOWER(title) AS lower_title, CHAR_LENGTH(title) AS title_length FROM film;
```

---
## SUBSTRING — extract part of a string

`SUBSTRING(str, start_position, length)` — positions are 1-indexed.

```python
%%sql
-- Q4: Show each film's title, its first 3 characters using LEFT, and characters 3 through 7 using SUBSTRING — from film table

SELECT title, LEFT(title, 3), SUBSTRING(title, 3, 5) FROM film;
```

---
## REPLACE — swap characters in a string

```python
%%sql
-- Q5: Show each customer's email with '@sakilacustomer.org' replaced by '@sakila.com' — from customer table — label the new column new_email

SELECT REPLACE(email, '@sakilacustomer.org', '@sakila.com') AS new_email FROM customer;
```

---
## TRIM — remove spaces

```python
%%sql
-- Q6: Strip spaces from the literal '   hello world   ' using TRIM, LTRIM, and RTRIM — show all three results in one row as trimmed, ltrimmed, rtrimmed

SELECT TRIM('   hello world   ') AS trimmed, LTRIM('   hello world   ') AS ltrimmed, RTRIM('   hello world   ') AS rtrimmed;
```

---
## INSTR — find position of a substring

Returns 0 if not found.

```python
%%sql
-- Q7: Show each film title and the position of 'LOVE' in the title using INSTR — from film table — only show rows where LOVE appears (position > 0)

SELECT title, INSTR(title,'LOVE') AS position FROM film WHERE INSTR(title,'LOVE') > 0;
```

## Practice Challenges

1. Show each actor's full name in 'LAST, First' format (last name uppercase)
2. Find all films whose title is longer than 20 characters
3. Show customer emails with the domain part extracted (everything after '@')
4. Find films that contain the word 'LOVE' anywhere in the title

```python
%%sql
-- Q8: Show each actor's name formatted as 'LAST, First' (last name uppercase, first name as-is) — from actor table — label the column formatted_name

SELECT CONCAT_WS(', ', UPPER(last_name), first_name) AS formatted_name FROM actor;
```

```python
%%sql
-- Q9: Get all films whose title is longer than 20 characters — from film table — show title and title_length

SELECT title, LENGTH(title) AS title_length FROM film WHERE CHAR_LENGTH(title) > 20;
```

```python
%%sql
-- Q10: Extract the domain from each customer's email (everything after '@') using SUBSTRING and INSTR — from customer table — show email and domain

SELECT email, SUBSTRING(email, INSTR(email, '@') + 1) AS domain FROM customer;
```

```python
%%sql
-- Q11: Get all films whose title contains the word 'LOVE' anywhere — from film table — show title

SELECT title FROM film WHERE title LIKE '%LOVE%';
```

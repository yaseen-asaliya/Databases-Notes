# Phase 11 — CASE Statements

CASE is SQL's if/else. It returns a value based on conditions and can be used anywhere an expression is valid — SELECT, WHERE, ORDER BY, GROUP BY.

**Searched CASE** (most common):
```sql
CASE
  WHEN condition1 THEN result1
  WHEN condition2 THEN result2
  ELSE default_result
END
```

**Simple CASE** (equality checks only):
```sql
CASE column
  WHEN 'value1' THEN result1
  WHEN 'value2' THEN result2
  ELSE default_result
END
```

- `ELSE` is optional — if omitted and no condition matches, returns NULL
- Always alias the CASE expression with `AS`

## Simple CASE — label film ratings

````python
%%sql
SELECT title, rating,
  CASE rating
    WHEN 'G'     THEN 'All ages'
    WHEN 'PG'    THEN 'Parental guidance'
    WHEN 'PG-13' THEN 'Parents strongly cautioned'
    WHEN 'R'     THEN 'Restricted'
    WHEN 'NC-17' THEN 'Adults only'
  END AS rating_description
FROM film
LIMIT 10;
````

---
## Searched CASE — bucket rental rates into price tiers

````python
%%sql
SELECT title, rental_rate,
  CASE
    WHEN rental_rate < 1.00 THEN 'Budget'
    WHEN rental_rate < 3.00 THEN 'Standard'
    ELSE 'Premium'
  END AS price_tier
FROM film
ORDER BY rental_rate
LIMIT 15;
````

---
## CASE inside COUNT — conditional aggregation

Count films per price tier without a subquery.

````python
%%sql
SELECT
  COUNT(CASE WHEN rental_rate < 1.00 THEN 1 END) AS budget,
  COUNT(CASE WHEN rental_rate BETWEEN 1.00 AND 2.99 THEN 1 END) AS standard,
  COUNT(CASE WHEN rental_rate >= 3.00 THEN 1 END) AS premium
FROM film;
````

---
## CASE in ORDER BY — custom sort order

Sort ratings in a specific business-defined order, not alphabetical.

````python
%%sql
SELECT title, rating
FROM film
ORDER BY
  CASE rating
    WHEN 'G'     THEN 1
    WHEN 'PG'    THEN 2
    WHEN 'PG-13' THEN 3
    WHEN 'R'     THEN 4
    WHEN 'NC-17' THEN 5
  END
LIMIT 10;
````

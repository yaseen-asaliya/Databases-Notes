# Phase 5 — CASE Statements

CASE is SQL's if/else. It returns a value based on conditions and can be used anywhere an expression is valid — SELECT, WHERE, ORDER BY, GROUP BY.

**Searched CASE** (most common):
```sql
CASE
  WHEN condition1 THEN result1
  WHEN condition2 THEN result2
  ELSE default_result
END
```

**Simple CASE** (equality checks only):
```sql
CASE column
  WHEN 'value1' THEN result1
  WHEN 'value2' THEN result2
  ELSE default_result
END
```

- `ELSE` is optional — if omitted and no condition matches, returns NULL
- Always alias the CASE expression with `AS`

## Simple CASE — label film ratings

````python
%%sql
-- Q1: Label each film's rating with a description using simple CASE — 'G'='General Audiences', 'PG'='Parental Guidance', 'PG-13'='Parents Strongly Cautioned', 'R'='Restricted', 'NC-17'='Adults Only' — from film table — show title, rating, rating_description

SELECT 
    title, 
    rating, 
    CASE rating
        WHEN 'G' THEN 'General Audiences'
        WHEN 'PG' THEN 'Parental Guidance'
        WHEN 'PG-13' THEN 'Parents Strongly Cautioned'
        WHEN 'R' THEN 'Restricted'
        WHEN 'NC-17' THEN 'Adults Only'
        ELSE 'Unknown'
    END AS rating_description
FROM film;
````

## Searched CASE — bucket rental rates into price tiers

````python
%%sql
-- Q2: Label each film's rental_rate as 'Budget' (0.99), 'Standard' (2.99), or 'Premium' (4.99) using a searched CASE — from film table — show title, rental_rate, and price_tier

SELECT 
    title, 
    rental_rate,
    CASE rental_rate
        WHEN 0.99 THEN 'Budget'
        WHEN 2.99 THEN 'Standard'
        WHEN 4.99 THEN 'Premium'
        ELSE 'Unknown'
    END AS price_tier
FROM film;
````

---
## CASE inside COUNT — conditional aggregation

Count films per price tier without a subquery.

````python
%%sql
-- Q3: Count how many films fall into each price tier using CASE inside COUNT — from film table — show three columns: budget_count, standard_count, premium_count — all in a single row

SELECT 
    COUNT(CASE rental_rate WHEN 0.99 THEN 1 END) AS budget_count, 
    COUNT(CASE rental_rate WHEN 2.99 THEN 1 END) AS standard_count,
    COUNT(CASE rental_rate WHEN 4.99 THEN 1 END) AS premium_count
FROM film;
````

## CASE in ORDER BY — custom sort order

Sort ratings in a specific business-defined order, not alphabetical.

````python
%%sql
-- Q4: List all films sorted by rating in this custom order: G first, then PG, PG-13, R, NC-17 — use CASE in ORDER BY — from film table — show title and rating

SELECT 
    title,
    rating
FROM film
ORDER BY 
    CASE rating
        WHEN 'G'     THEN 1
        WHEN 'PG'    THEN 2
        WHEN 'PG-13' THEN 3
        WHEN 'R'     THEN 4
        WHEN 'NC-17' THEN 5
    END;
````

## Practice Challenges

1. Label each film's length as 'Short' (< 60 min), 'Medium' (60–120), or 'Long' (> 120)
2. Count how many customers are active vs inactive using CASE inside COUNT
3. Show each payment with an 'amount_category': 'Low' < 2, 'Medium' 2–5, 'High' > 5

````python
%%sql
-- Q5: Label each film's length as 'Short' (under 60 min), 'Medium' (60–120 min), or 'Long' (over 120 min) — from film table — show title, length, and length_category

SELECT 
    title, 
    length,
    CASE 
        WHEN length < 60 THEN 'Short'
        WHEN length >= 60 AND length <= 120 THEN 'Medium'
        WHEN length > 120 THEN 'Long'       
        ELSE 'Unknown'
    END AS length_category
FROM film;
````

````python
%%sql
-- Q6: Count how many customers are active vs inactive using CASE inside COUNT — from customer table — show active_count and inactive_count in a single row

SELECT 
    COUNT(CASE active WHEN 1 THEN 1 END) AS active_count,
    COUNT(CASE active WHEN 0 THEN 1 END) AS inactive_count
FROM customer;
````

````python
%%sql
-- Q7: Label each payment with an amount_category: 'Low' for amount < 2, 'Medium' for 2–5, 'High' for above 5 — from payment table — show payment_id, amount, and amount_category

SELECT 
    payment_id, 
    amount,
    CASE 
        WHEN amount < 2 THEN 'Low'
        WHEN amount >= 2 AND amount <= 5 THEN 'Medium'
        WHEN amount > 5 THEN 'High'
        ELSE 'Unknown'
    END AS amount_category
FROM payment; 
````

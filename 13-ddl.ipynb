# Phase 16 — DDL (Data Definition Language)

DDL is the set of SQL commands that define and modify **structure** (not data).

| Command | What it does |
|---|---|
| `CREATE TABLE` | Define a new table |
| `ALTER TABLE` | Modify an existing table |
| `DROP TABLE` | Delete a table and all its data |
| `TRUNCATE TABLE` | Delete all rows, keep structure (faster than DELETE) |

## Common Data Types

| Type | Use for |
|---|---|
| `INT` / `BIGINT` | Integers |
| `DECIMAL(p, s)` | Exact decimals — money, measurements |
| `FLOAT` / `DOUBLE` | Approximate decimals — avoid for money |
| `VARCHAR(n)` | Variable-length string up to n chars |
| `CHAR(n)` | Fixed-length string — wastes space but faster for short fixed values |
| `TEXT` | Long text, no length limit |
| `DATE` | Date only (YYYY-MM-DD) |
| `DATETIME` | Date + time |
| `TIMESTAMP` | Like DATETIME but stores UTC; auto-updates available |
| `BOOLEAN` | TRUE/FALSE (stored as TINYINT 1/0) |
| `ENUM(...)` | One value from a fixed list |

## Constraints

| Constraint | Enforces |
|---|---|
| `PRIMARY KEY` | Unique + NOT NULL identifier |
| `NOT NULL` | Column must always have a value |
| `UNIQUE` | No duplicates allowed |
| `DEFAULT value` | Value used when none is provided |
| `FOREIGN KEY` | Value must exist in referenced table |
| `CHECK` | Custom condition must be true |

## CREATE TABLE with constraints

```python
%%sql
CREATE TABLE IF NOT EXISTS review (
  review_id   INT           AUTO_INCREMENT PRIMARY KEY,
  film_id     SMALLINT UNSIGNED NOT NULL,
  customer_id SMALLINT UNSIGNED NOT NULL,
  rating      TINYINT       NOT NULL,
  body        TEXT,
  created_at  DATETIME      DEFAULT CURRENT_TIMESTAMP,
  CONSTRAINT fk_review_film     FOREIGN KEY (film_id)     REFERENCES film(film_id),
  CONSTRAINT fk_review_customer FOREIGN KEY (customer_id) REFERENCES customer(customer_id),
  CONSTRAINT chk_rating CHECK (rating BETWEEN 1 AND 5)
);
```

## ALTER TABLE — add, modify, drop columns

```python
%%sql
-- Add a column
ALTER TABLE review ADD COLUMN helpful_votes INT DEFAULT 0;
```

```python
%%sql
-- Rename a column
ALTER TABLE review RENAME COLUMN body TO review_text;
```

```python
%%sql
-- Change a column's type
ALTER TABLE review MODIFY COLUMN helpful_votes SMALLINT DEFAULT 0;
```

```python
%%sql
-- Drop a column
ALTER TABLE review DROP COLUMN helpful_votes;
```

---
## DROP TABLE vs TRUNCATE TABLE

- `DROP TABLE` removes the table entirely (structure + data)
- `TRUNCATE TABLE` removes all rows but keeps the table structure; resets AUTO_INCREMENT

```python
%%sql
DROP TABLE IF EXISTS review;
```

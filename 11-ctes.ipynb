# Phase 10 — CTEs (Common Table Expressions)

A CTE is a named temporary result defined at the top of a query with `WITH`. It makes complex queries readable by breaking them into named steps.

```sql
WITH cte_name AS (
    SELECT ...
)
SELECT * FROM cte_name;
```

**CTE vs Subquery:** same result, but CTEs are reusable within the same query and easier to read.

**Multiple CTEs:** chain with commas — each can reference the ones before it.
```sql
WITH
  cte1 AS (SELECT ...),
  cte2 AS (SELECT ... FROM cte1)
SELECT * FROM cte2;
```

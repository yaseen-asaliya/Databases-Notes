# Phase 19 — Stored Procedures

A stored procedure is reusable SQL logic saved in the database, called by name.

```sql
CREATE PROCEDURE procedure_name (IN param1 TYPE, OUT param2 TYPE)
BEGIN
  -- SQL statements
END;

CALL procedure_name(arg1, @out_var);
```

**Parameter modes:**
| Mode | Direction |
|------|----------|
| `IN` | Caller passes a value in |
| `OUT` | Procedure writes a value back to the caller |
| `INOUT` | Both directions |

**Stored procedure vs function:**
- Procedure: called with `CALL`, can have OUT params, cannot be used inside SELECT
- Function: called inside SELECT/WHERE like any expression, must return a single value

**Important:** `DELIMITER` must be changed when creating procedures in the CLI (not needed in notebooks — each statement is sent separately via SQLAlchemy).

**Why use stored procedures:**
- Encapsulate complex multi-step logic
- Reduce network round-trips (one CALL vs many queries)
- Security: grant EXECUTE on procedure without exposing underlying tables

## Create a simple procedure

**Note:** Creating procedures with BEGIN...END requires loading them via the MySQL CLI (DELIMITER issue). The cell below shows the syntax — run it via:
```bash
docker exec -it sql-practice mysql -u root -ppractice sakila
```

```sql
CREATE PROCEDURE top_customers(IN n INT)
BEGIN
  SELECT c.first_name, c.last_name, SUM(p.amount) AS total_spent
  FROM payment p
  JOIN customer c ON p.customer_id = c.customer_id
  GROUP BY p.customer_id
  ORDER BY total_spent DESC
  LIMIT n;
END;
```

Then call it:
```sql
CALL top_customers(5);
```

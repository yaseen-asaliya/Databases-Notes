# Phase 14 — UNION / UNION ALL

Combines result sets from two or more SELECT statements into one.

```sql
SELECT col1, col2 FROM table_a
UNION
SELECT col1, col2 FROM table_b;
```

**Rules:**
- Both queries must return the **same number of columns**
- Columns are matched by **position**, not name
- The column names in the result come from the **first** SELECT

| Operator | Duplicates |
|---|---|
| `UNION` | Removes duplicates (slower — does a distinct pass) |
| `UNION ALL` | Keeps all rows including duplicates (faster) |

**When to use UNION vs JOIN:**
- JOIN → combine columns from related tables (horizontal)
- UNION → stack rows from similar queries (vertical)

## UNION — combine staff and customer names into one list

Useful for finding all people in the system regardless of role.

````python
%%sql
SELECT first_name, last_name, 'customer' AS role FROM customer
UNION
SELECT first_name, last_name, 'staff'    AS role FROM staff
ORDER BY last_name
LIMIT 15;
````

---
## UNION ALL — keep duplicates (faster)

If the same person were in both tables, UNION would deduplicate, UNION ALL would show them twice.

````python
%%sql
-- Count rows: UNION ALL keeps everything
SELECT 'customers' AS source, COUNT(*) AS total FROM customer
UNION ALL
SELECT 'staff',               COUNT(*) AS total FROM staff;
````

---
## UNION with different conditions on the same table

Get a combined list: films under $1 and films over $4 (the extremes).

````python
%%sql
SELECT title, rental_rate, 'cheap' AS tier
FROM film WHERE rental_rate < 1.00
UNION ALL
SELECT title, rental_rate, 'expensive'
FROM film WHERE rental_rate > 4.00
ORDER BY rental_rate
LIMIT 15;
````

---
## ORDER BY and LIMIT with UNION

`ORDER BY` and `LIMIT` after a UNION apply to the **combined** result. To sort/limit individual parts, wrap them in subqueries.

````python
%%sql
-- Top 3 customers + top 3 staff by name
SELECT * FROM (
    SELECT first_name, last_name, 'customer' AS role FROM customer ORDER BY last_name LIMIT 3
) a
UNION ALL
SELECT * FROM (
    SELECT first_name, last_name, 'staff' AS role FROM staff ORDER BY last_name LIMIT 3
) b;
````
